# 03. Stromlinien

Stromlinien verlaufen tangential zum jeweiligen Geschwindigkeitsfeld. Hinter der Stufe kann sich eine Rezirkulationszone ausbilden.

**Interaktion:** Über den Regler **Stufenhöhe h** lässt sich die rückwärtsgerichtete Stufe verändern. Die Strömung wird aus den Anfangsbedingungen neu berechnet. Die zeitliche Entwicklung ist animiert; mit **Pause**, **Fortsetzen** und **Neu starten** kann sie vorgeführt werden.

Diese Seite enthält **nur diese eine Darstellung**.

```python
%matplotlib widget
from s01_parameter import Parameter
from s06_simulation import Simulation
from s08_einzelseite import InteraktiveEinzelseite

# Diese Präsentationsseite zeigt ausschließlich: Stromlinien
parameter = Parameter().with_h(1.0)
simulation = Simulation(parameter)
seite = InteraktiveEinzelseite(simulation, seitennummer=3,
                              schritte_pro_bild=8, intervall=170)
seite.anzeigen()
```

### Erklärung zum Rechenmodell

Das versetzte **MAC-Gitter** unterscheidet Zellmittelpunkte für den Druck und Zellflächen für die Geschwindigkeiten. Der explizite Impulsschritt liefert die Zwischengeschwindigkeiten **F** und **G**. Anschließend wird die **Druck-Poisson-Gleichung** gelöst und die Geschwindigkeit projiziert. Bei einer Änderung von **h** entsteht eine neue Fluidmaske und eine neue Druckmatrix.

Die bereitgestellten Vorlesungswerte umfassen **Re = 100**, **α = 0,9**, **τ = 0,5**, **ω = 1,7**, **ε = 0,001**, **imax = 100**, **jmax = 20**. Der Standarddrucklöser verwendet für schnelle Live-Vorführungen eine LU-Zerlegung; alternativ ist SOR verfügbar.
